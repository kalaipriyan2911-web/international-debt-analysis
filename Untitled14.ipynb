````python
# ============================================================
# 🌍 INTERNATIONAL DEBT ANALYSIS SYSTEM
# STEP 10: FORWARD FILL AND BACKWARD FILL COMPARISON
# ============================================================

import pandas as pd
from IPython.display import display, Markdown

# Display project heading
display(Markdown("# 🌍 International Debt Analysis System"))
display(Markdown("## 🔄 Forward Fill and Backward Fill Comparison"))

# ------------------------------------------------------------
# STEP 1: Identify year columns
# ------------------------------------------------------------

# Select columns containing year values, such as 2000, 2001, etc.
year_columns = [
    col for col in debt.columns
    if str(col).isdigit() and len(str(col)) == 4
]

print("Year columns identified:", len(year_columns))

# ------------------------------------------------------------
# STEP 2: Create separate copies for testing
# ------------------------------------------------------------

# Preserve the original dataset without making changes
debt_original = debt.copy()

# Create a copy for Forward Fill testing
debt_ffill_test = debt.copy()

# Create a copy for Backward Fill testing
debt_bfill_test = debt.copy()

# ------------------------------------------------------------
# STEP 3: Apply Forward Fill to year columns for comparison
# ------------------------------------------------------------

# Fill missing values using the previous year's available value
# This is a preliminary test, not the final cleaned dataset
debt_ffill_test[year_columns] = (
    debt_ffill_test[year_columns].ffill(axis=1)
)

# ------------------------------------------------------------
# STEP 4: Apply Backward Fill to year columns for comparison
# ------------------------------------------------------------

# Fill missing values using the next year's available value
# This is a preliminary test, not the final cleaned dataset
debt_bfill_test[year_columns] = (
    debt_bfill_test[year_columns].bfill(axis=1)
)

# ------------------------------------------------------------
# STEP 5: Compare missing values
# ------------------------------------------------------------

original_missing = debt_original[year_columns].isnull().sum().sum()

ffill_missing = debt_ffill_test[year_columns].isnull().sum().sum()

bfill_missing = debt_bfill_test[year_columns].isnull().sum().sum()

# ------------------------------------------------------------
# STEP 6: Display the comparison report
# ------------------------------------------------------------

comparison_report = pd.DataFrame({
    "Method": [
        "Original Data",
        "Forward Fill Test",
        "Backward Fill Test"
    ],
    "Remaining Missing Values": [
        original_missing,
        ffill_missing,
        bfill_missing
    ]
})

display(Markdown("### 📊 Missing Values Comparison"))
display(comparison_report)

print("\nOriginal dataset preserved.")
print("Comparison completed successfully!")
````

````python
# ============================================================
# 🌍 INTERNATIONAL DEBT ANALYSIS SYSTEM
# STEP 12: DATA CLEANING
# ============================================================

import pandas as pd
import numpy as np
from IPython.display import display, Markdown

# ------------------------------------------------------------
# STEP 1: Display project heading
# ------------------------------------------------------------

display(Markdown("# 🌍 International Debt Analysis System"))
display(Markdown("## 🧹 Step 12: Data Cleaning"))

# ------------------------------------------------------------
# STEP 2: Create a copy of the original debt dataset
# ------------------------------------------------------------

# Keep the original debt dataset unchanged
debt_cleaned = debt.copy()

# Store the original row count
original_rows = len(debt_cleaned)

# Count duplicate rows before cleaning
duplicate_rows = debt_cleaned.duplicated().sum()

print("Original Rows:", original_rows)
print("Duplicate Rows Found:", duplicate_rows)

# ------------------------------------------------------------
# STEP 3: Remove exact duplicate rows
# ------------------------------------------------------------

# Remove only fully duplicated records
debt_cleaned = debt_cleaned.drop_duplicates().copy()

# ------------------------------------------------------------
# STEP 4: Clean column names
# ------------------------------------------------------------

# Remove unnecessary spaces around column names
debt_cleaned.columns = debt_cleaned.columns.str.strip()

# ------------------------------------------------------------
# STEP 5: Clean text values
# ------------------------------------------------------------

# Remove unnecessary spaces from text columns
text_columns = debt_cleaned.select_dtypes(
    include=["object", "string"]
).columns

for column in text_columns:
    debt_cleaned[column] = debt_cleaned[column].apply(
        lambda value: value.strip()
        if isinstance(value, str)
        else value
    )

# ------------------------------------------------------------
# STEP 6: Display the cleaning summary
# ------------------------------------------------------------

display(Markdown("### 📊 Data Cleaning Summary"))

print("Rows Before Cleaning:", original_rows)
print("Rows After Cleaning:", len(debt_cleaned))
print("Duplicate Rows Removed:", original_rows - len(debt_cleaned))
print("Columns:", len(debt_cleaned.columns))
print(
    "Missing Values Remaining:",
    debt_cleaned.isnull().sum().sum()
)

print("\nData cleaning completed successfully!")
````

````python
# ============================================================
# 🌍 INTERNATIONAL DEBT ANALYSIS SYSTEM
# STEP 13: CLEAN REMAINING DATASETS
# ============================================================

from IPython.display import display, Markdown

# Display project heading
display(Markdown("# 🌍 International Debt Analysis System"))
display(Markdown("## 🧹 Cleaning Remaining Datasets"))

# ------------------------------------------------------------
# STEP 1: Create copies of the original datasets
# ------------------------------------------------------------

country_cleaned = country.copy()
indicator_cleaned = indicator.copy()
footnotes_cleaned = footnotes.copy()
country_series_cleaned = country_series.copy()

# ------------------------------------------------------------
# STEP 2: Create a dictionary of datasets to clean
# ------------------------------------------------------------

datasets_to_clean = {
    "Country Metadata": country_cleaned,
    "Indicator Metadata": indicator_cleaned,
    "Footnotes": footnotes_cleaned,
    "Country Series Metadata": country_series_cleaned
}

# ------------------------------------------------------------
# STEP 3: Clean each dataset
# ------------------------------------------------------------

cleaned_datasets = {}

for name, df in datasets_to_clean.items():

    # Remove unnecessary spaces from column names
    df.columns = df.columns.str.strip()

    # Remove unnecessary spaces from text values
    text_columns = df.select_dtypes(
        include=["object", "string"]
    ).columns

    for column in text_columns:
        df[column] = df[column].apply(
            lambda value: value.strip()
            if isinstance(value, str)
            else value
        )

    # Remove exact duplicate rows only
    before_rows = len(df)
    df = df.drop_duplicates().copy()
    duplicates_removed = before_rows - len(df)

    # Store the cleaned dataset
    cleaned_datasets[name] = df

    # Display cleaning results
    print("\n" + "=" * 50)
    print("Dataset:", name)
    print("Rows Before Cleaning:", before_rows)
    print("Rows After Cleaning:", len(df))
    print("Duplicate Rows Removed:", duplicates_removed)
    print("Missing Values Remaining:", df.isnull().sum().sum())

# Update cleaned dataset variables
country_cleaned = cleaned_datasets["Country Metadata"]
indicator_cleaned = cleaned_datasets["Indicator Metadata"]
footnotes_cleaned = cleaned_datasets["Footnotes"]
country_series_cleaned = cleaned_datasets["Country Series Metadata"]

print("\nAll remaining datasets cleaned successfully!")
````

````python
# ==========================================
# STEP 30: FIX MISSING VALUES FOR SQL IMPORT
# ==========================================

import pandas as pd
import numpy as np

# Load the SQL-ready CSV file
debt_final = pd.read_csv(
    "Debt_Data_SQL_Ready.csv",
    low_memory=False
)

# Identify the year columns
year_columns = [
    col for col in debt_final.columns
    if col.isdigit()
]

# Convert year columns to numeric values
# Invalid values such as '?' become NaN
for col in year_columns:
    debt_final[col] = pd.to_numeric(
        debt_final[col].replace(
            ["?", "NA", "N/A", ""],
            np.nan
        ),
        errors="coerce"
    )

# Save missing numeric values as empty CSV cells
# Do not fill them with artificial debt values
debt_final.to_csv(
    "Debt_Data_SQL_Ready_Fixed.csv",
    index=False,
    na_rep=""
)

print("✅ Corrected CSV saved successfully!")
print("Total rows:", len(debt_final))
print("Total columns:", len(debt_final.columns))
print(
    "Missing values in year columns:",
    debt_final[year_columns].isna().sum().sum()
)
````

````python
import pandas as pd
import numpy as np

debt = pd.read_csv("IDS_ALLCountries_Data (1).csv", encoding="latin1")
country = pd.read_csv("IDS_CountryMetaData (2).csv", encoding="latin1")
indicator = pd.read_csv("IDS_SeriesMetaData (2).csv", encoding="latin1")
footnotes = pd.read_csv("IDS_FootNoteMetaData (1).csv", encoding="latin1")
country_series = pd.read_csv("Country-Series - Metadata (1).csv", encoding="latin1")

print("Debt:", debt.shape)
print("Country:", country.shape)
print("Indicator:", indicator.shape)
print("Footnotes:", footnotes.shape)
print("Country-Series:", country_series.shape)
````

````python
# Check duplicate rows
print("Duplicate rows:", debt.duplicated().sum())

# Remove duplicate rows
debt = debt.drop_duplicates().copy()

# Check missing values
print("Total missing values:", debt.isnull().sum().sum())

# Check dataset size after cleaning
print("Cleaned dataset shape:", debt.shape)

# Display first 5 records
display(debt.head())
````

````python
# 1. Country-wise record count
country_count = debt.groupby("Country Name").size().sort_values(ascending=False)

print("Top 10 Countries by Number of Records:")
display(country_count.head(10).reset_index(name="Record Count"))

# 2. Debt indicators with most records
indicator_count = debt.groupby("Series Name").size().sort_values(ascending=False)

print("Top 10 Debt Indicators:")
display(indicator_count.head(10).reset_index(name="Record Count"))

# 3. Year-wise missing values
year_columns = [col for col in debt.columns if str(col).isdigit()]

print("Number of year columns:", len(year_columns))
print("\nMissing values in recent years:")
display(debt[["Country Name"] + [col for col in ["2020", "2021", "2022", "2023", "2024"] if col in debt.columns]].isnull().sum().reset_index(name="Missing Count"))
````

````python
# Save cleaned dataset
debt.to_csv("International_Debt_Cleaned.csv", index=False)

# Save country-wise analysis
country_count.head(10).reset_index(name="Record Count").to_csv(
    "Top_10_Countries.csv", index=False
)

# Save indicator analysis
indicator_count.head(10).reset_index(name="Record Count").to_csv(
    "Top_10_Debt_Indicators.csv", index=False
)

print("All 3 CSV files saved successfully!")
print("Cleaned dataset shape:", debt.shape)
````
